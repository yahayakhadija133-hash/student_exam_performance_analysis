```python
X = df[["Study_Hours_Per_Week",
        "Attendance_Percentage",
        "Previous_Score",
        "Sleep_Hours"]]

y = df["Pass_Fail"]

X.head()

y.head()
```

```python
we need split data we need to divide data into two part 
tranining data , used to teach the computer 
testing data used to test what the computer has learn
```

```python
from sklearn.model_selection import train_test_split
```

```python
#a  machine learning library import atool from sciket_learn
```

```python
#This tells Python:

"Split my dataset into training data and testing data."
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
#(7, 4) means:

7 rows (students) traning student with four features

4 columns (features)

(3, 4) means:3 student wiyh four feature

3 rows (students)
4 columns (features)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
import decision making tree algorithm is a machine learning  model that 
makes decision like a flowchat
```

```python
#This creates the machine learning model.

Think of it as preparing an empty brain that is ready to learn.
```

```python
model = DecisionTreeClassifier(random_state=42)

model.fit(X_train, y_train)
```

```python
fit() means:

Teach the model using the training data.

The model studies the 7 training students and learns patterns that
separate Pass from Fail.
```

```python
predictions = model.predict(X_test)

print(predictions)
#Use what you learned to predict whether the test students will Pass or Fail."

#The output might look like: prediction of twostudent in the test set
```

```python
#This means your machine learning model predicted:

#Student 1 → Fail
#Student 2 → Pass
#Student 3 → Pass

#Now we need to answer an important question:

#"Are these predictions correct?"

#Compare the Predictions with the Actual Results
```

```python
print("Predictions:")
print(predictions)

print("\nActual:")
print(y_test.values)
```

```python
# It prints:

The model's predictions.
The actual answers from your dataset.

For example:
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, predictions)

print("Accuracy:", accuracy)
```

```python
# Accuracy tells us:

Out of all the students in the test set, how many did the model predict correctly?

Example

Suppose there are 3 students:

Student	Actual	Predicted
1	Pass	Pass ✅
2	Fail	Fail ✅
3	Pass	Fail ❌

The model got 2 out of 3 correct.

Accuracy is:

3
2
	​

=0.67

which means 67% accuracy.

If it gets all 3 correct, the accuracy is: is 1.0 ie 100% accuracy
```

```python
# Why did you get 100%?

Because your dataset has only 10 students.

When you split it:

Training data = 7 students
Testing data = 3 students

The model predicted all 3 correctly
```

```python
# Does that mean the model is perfect?

No.
    "The model achieved 100% accuracy because the dataset is very small (10 records).
    With a larger dataset, the accuracy may be lower 
    but would provide a more realistic evaluation."
```

```python
Confusion Matrix

A Confusion Matrix is a table that compares:

Actual results (the true Pass/Fail)
Predicted results (what the model guessed)

It helps us see exactly where the model was correct or wrong.
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, predictions)

print(cm)
```

```python
#This imports the Confusion Matrix tool from scikit-learn.

Think of it as bringing a calculator into your notebook.
    y_test = the real answers
predictions = the model's answers
and stores the result in a variable called cm.
               Predicted
             Fail   Pass

Actual Fail    1      0

Actual Pass    0      2
    Meaning:

1 Fail student was correctly predicted as Fail. ✅
2 Pass students were correctly predicted as Pass. ✅
No mistakes were made.

That is why your accuracy was 100%.
```

```python
# adding a section called predicting a new student
this is not an error but show that i can apply a new data to my dataset which
is not in my existing record

This happens because:

During training, you used a DataFrame with column names:
Study_Hours_Per_Week
Attendance_Percentage
Previous_Score
Sleep_Hours
But when predicting, you gave the model a simple Python list:
```

```python

new_student = [[15, 90, 85, 8]]
The new student has:

Study Hours = 15
Attendance = 90%
Previous Score = 85
Sleep Hours = 8

The model predicts whether that student will Pass or Fail.

This show model can predict new data not just analysis existig record
```

```python
The model still makes the correct prediction, but it warns you that 
the new data has no column names.
```

```python
How to remove the warning 

Instead of using a list, use a DataFrame:
```

```python
new_student = pd.DataFrame({
    "Study_Hours_Per_Week": [15],
    "Attendance_Percentage": [90],
    "Previous_Score": [85],
    "Sleep_Hours": [8]
})

prediction = model.predict(new_student)

print("Prediction:", prediction)
```
